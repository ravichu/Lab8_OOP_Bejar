EXERCISE 1

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.available = True


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author):
        self.books.append(Book(title, author))
        print(f"Book '{title}' added.")

    def borrow_book(self, title):
        for book in self.books:
            if book.title == title:
                if book.available:
                    book.available = False
                    print(f"You borrowed '{title}'.")
                else:
                    print(f"'{title}' is already borrowed.")
                return
        print("Book not found.")

    def return_book(self, title):
        for book in self.books:
            if book.title == title:
                book.available = True
                print(f"You returned '{title}'.")
                return
        print("Book not found.")

    def show_books(self):
        for book in self.books:
            status = "Available" if book.available else "Borrowed"
            print(f"{book.title} by {book.author} - {status}")
```

```python
library = Library()

library.add_book("Python Basics", "John Doe")
library.add_book("OOP Concepts", "Jane Smith")

library.show_books()

library.borrow_book("Python Basics")
library.show_books()

library.return_book("Python Basics")
library.show_books()
```

```python
class Employee:
    def __init__(self, name, hourly_rate, hours_worked):
        self.name = name
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_salary(self):
        if self.hours_worked > 160:
            overtime = self.hours_worked - 160
            overtime_pay = overtime * self.hourly_rate * 1.5
            base_pay = 160 * self.hourly_rate
        else:
            overtime_pay = 0
            base_pay = self.hours_worked * self.hourly_rate

        total_salary = base_pay + overtime_pay
        return base_pay, overtime_pay, total_salary
```

```python
employees = [
    Employee("Alice", 100, 170),
    Employee("Bob", 120, 150)
]

for emp in employees:
    base, overtime, total = emp.calculate_salary()
    print(f"\nName: {emp.name}")
    print(f"Hours Worked: {emp.hours_worked}")
    print(f"Base Salary: {base}")
    print(f"Overtime Pay: {overtime}")
    print(f"Total Salary: {total}")
```

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock


class ShoppingCart:
    def __init__(self):
        self.cart = []

    def add_product(self, product, quantity):
        if product.stock >= quantity:
            self.cart.append((product, quantity))
            product.stock -= quantity
            print(f"Added {quantity} {product.name}(s) to cart.")
        else:
            print("Not enough stock.")

    def remove_product(self, product_name):
        for item in self.cart:
            if item[0].name == product_name:
                self.cart.remove(item)
                print(f"Removed {product_name} from cart.")
                return
        print("Product not found in cart.")

    def show_cart(self):
        total = 0
        for product, quantity in self.cart:
            cost = product.price * quantity
            total += cost
            print(f"{product.name} x{quantity} = {cost}")
        print("Total:", total)
```

```python
p1 = Product("Laptop", 50000, 5)
p2 = Product("Mouse", 500, 10)

cart = ShoppingCart()

cart.add_product(p1, 1)
cart.add_product(p2, 2)

cart.show_cart()

cart.remove_product("Mouse")
cart.show_cart()
```
