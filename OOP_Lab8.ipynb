```python
# Encapsulation

class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        return f"New balance: {self.__balance}"

    def get_balance(self):
        return f"Balance: {self.__balance}"

account = BankAccount("Alice", 5000)

print(account.deposit(1500))
print(account.get_balance())
```

```python
# Polymorphism

class Shape:
    def area(self):
        return "Calculating..."

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, l, w):
        self.l = l
        self.w = w

    def area(self):
        return self.l * self.w

c = Circle(5)
r = Rectangle(4, 6)

print("Circle:", c.area())
print("Rectangle:", r.area())
```

```python
# Operator Overloading

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def __add__(self, other):
        return BankAccount(self.balance + other.balance)

account1 = BankAccount(5000)
account2 = BankAccount(3000)

total = account1 + account2

print("Total Balance:", total.balance)
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def display(self):
        return f"({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(1, 1)

v_add = v1 + v2
v_sub = v1 - v2

print("Addition:", v_add.display())
print("Subtraction:", v_sub.display())
```

```python
# MRO Example

class A:
    def show(self):
        return "Class A"

class B(A):
    def show(self):
        return "Class B"

class C(A):
    def show(self):
        return "Class C"

class D(B, C):
    pass

obj = D()

print(obj.show())
```

```python
class ATM:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient balance"
        
        self.balance -= amount
        return f"Withdrawn {amount}, Remaining balance: {self.balance}"

atm = ATM(1000)
print(atm.withdraw(200))
```

ADVANCED TECHNIQUES

```python
from abc import ABC, abstractmethod
from dataclasses import dataclass

# --- 1. Abstract Base Classes ---
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Bark! Bark!"

# --- 2. Data Classes ---
@dataclass
class Student:
    name: str
    age: int
    course: str

# --- 3. Custom Exception Handling ---
class InvalidAgeError(Exception):
    pass

def set_age(age):
    if age < 0:
        raise InvalidAgeError("Age cannot be negative")
    return age

# --- 4. Property Decorators ---
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def fahrenheit(self):
        return (self._celsius * 9/5) + 32

# --- 5. Composition Over Inheritance ---
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        self.engine = Engine() # Composition: Car "has an" engine

    def drive(self):
        status = self.engine.start()
        return f"{status}... Car is moving"

# --- EXECUTION ---
print("--- Advanced Techniques Outputs ---")

# 1. ABC
dog = Dog()
print(f"ABC (Animal): {dog.sound()}")

# 2. Data Class
std = Student("Alice", 20, "CS")
print(f"Data Class: {std}")

# 3. Custom Exception
try:
    print(f"Age set to: {set_age(-5)}")
except InvalidAgeError as e:
    print(f"Custom Exception: {e}")

# 4. Property
temp = Temperature(25)
print(f"Property (Fahrenheit): {temp.fahrenheit}")

# 5. Composition
my_car = Car()
print(f"Composition: {my_car.drive()}")
```
